```python
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
titanic = sns.load_dataset("titanic")
```

```python
titanic
```

```python
sns.displot(data=titanic, x="fare", kde=True)
```

```python
sns.displot(data=titanic, x="fare", kind="ecdf")
```

```python
sns.displot(data=titanic, x="fare", y="age")
```

```python
sns.displot(data=titanic, x="fare", y="age", kind="kde")
```

```python
sns.displot(data=titanic, x="fare", y="age", kind="kde", rug=True)
```

```python
sns.displot(data=titanic, x="fare", hue="embark_town", kind="kde")
```

```python
sns.displot(data=titanic, x="fare", hue="embark_town", col="sex", kind="kde")
```

```python
sns.displot(data=titanic, y="age", hue="sex", col="embark_town",
            kind="kde", height=5, aspect=.7,
            )
```

```python
sns.displot(data=titanic, y="age", hue="sex", col="embark_town",
            kind="ecdf", height=5, aspect=.7,
            )
```
