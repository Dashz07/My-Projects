```python
import seaborn as sns
```

```python
penguins = sns.load_dataset('penguins')
```

```python
sns.barplot(penguins, x="island", y="body_mass_g")
```

```python
sns.barplot(penguins, y="island", x ="body_mass_g", hue = "island", legend = False)
```

```python
sns.barplot(penguins, x="island", y="body_mass_g", hue = "sex")
```

```python
sns.barplot(penguins, x="island", y="body_mass_g", errorbar="sd")
```

```python
sns.barplot(
    penguins,x="body_mass_g", y="island",
    errorbar=("pi", 50), capsize=.5,
    err_kws={"color": ".5", "linewidth": 2.5},
    linewidth=2.5, edgecolor=".5", facecolor=(0,0,0,0)
)
```

```python
sns.catplot(
    penguins, kind="bar",
    x="sex", y="body_mass_g", col="species",
    height=5, aspect=.5,
)
```

```python
sns.catplot(
    penguins, kind="bar",
    x="sex", y="body_mass_g", col="island", hue="species",
    height=5, aspect=.5,
)
```
