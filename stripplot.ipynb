```python
import seaborn as sns
```

```python
tips = sns.load_dataset("tips")
tips
```

```python
sns.stripplot(data=tips, x="total_bill")
```

```python
sns.stripplot(data=tips, x="total_bill",y="tip")
```

```python
sns.stripplot(data=tips, y="total_bill",x="tip")
```

```python
sns.stripplot(data=tips, x="total_bill",y="day")
```

```python
sns.stripplot(data=tips, y="total_bill",x="day")
```

```python
sns.stripplot(data=tips, x="total_bill",y="day",hue="day", legend=False)
```

```python
sns.stripplot(data=tips, x="total_bill",y="day", hue="sex" )
```

```python
sns.stripplot(data=tips, x="total_bill",y="size",hue="tip")
```

```python
sns.stripplot(data=tips, x="total_bill",y="day",hue="size")
```

```python
sns.stripplot(data=tips, x="total_bill",y="day",hue="size", palette="bright")
```

```python
sns.stripplot(data=tips, x="total_bill",y="size",hue="sex",dodge=True)
```

```python
sns.stripplot(data=tips, x="total_bill",y="day",hue="tip",dodge=True,jitter=False)
```

```python
sns.stripplot(data=tips)
```

```python
sns.stripplot(data=tips, orient="h")
```

```python
sns.stripplot(data=tips,x="total_bill",y="tip", orient="h", hue="smoker")
```

```python
sns.stripplot(
    data=tips.query("size in [3,4,6]"),
    x="total_bill",y="tip",orient="h")
```

```python
sns.stripplot(
    data=tips.query("size in [3,4,6]"),
    x="total_bill",y="tip",orient="h",
    native_scale=True,
    )
```

```python
sns.stripplot(
    data=tips, x="total_bill", y="size", hue="time", palette="bright",
    jitter=False, s=20 ,marker="4", linewidth=1, alpha=1,
    )
```
